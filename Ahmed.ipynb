```python
import numpy as np
import pandas as pd

from sklearn.linear_model import LogisticRegression
```

```python
X_train = np.load("X_train.npy")
y_train = np.load("y_train.npy")

X_val = np.load("X_val.npy")
y_val = np.load("y_val.npy")

X_test = np.load("X_test.npy")
y_test = np.load("y_test.npy")
```

## Logistic Regression

```python
logistic_model = LogisticRegression(random_state=42)
```

```python
logistic_model.fit(X_train, y_train)
```

```python
y_val_pred = logistic_model.predict(X_val)
```

```python
from sklearn.metrics import accuracy_score

val_accuracy = accuracy_score(y_val, y_val_pred)
print("Validation Accuracy:", val_accuracy)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_val, y_val_pred)
print(cm)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_val, y_val_pred))
```

## K-Nearest Neighbors (KNN)

```python
  from sklearn.neighbors import KNeighborsClassifier
```

```python
knn_model = KNeighborsClassifier(n_neighbors=5)
```

```python
knn_model.fit(X_train, y_train)
```

```python
y_val_pred_knn = knn_model.predict(X_val)
```

```python
knn_accuracy = accuracy_score(y_val, y_val_pred_knn)
print("KNN Validation Accuracy:", knn_accuracy)
```

```python
print(classification_report(y_val, y_val_pred_knn))
```

## Support Vector Machine (SVM)

```python
from sklearn.svm import SVC
```

```python
svm_model = SVC(kernel="rbf", random_state=42)
```

```python
svm_model.fit(X_train, y_train)
```

```python
y_val_pred_svm = svm_model.predict(X_val)
```

```python
svm_accuracy = accuracy_score(y_val, y_val_pred_svm)
print("SVM Validation Accuracy:", svm_accuracy)
```

```python
print(classification_report(y_val, y_val_pred_svm))
```

## Model Evaluation & Comparison

```python
results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "KNN",
        "SVM"
    ],
    "Validation Accuracy": [
        val_accuracy,
        knn_accuracy,
        svm_accuracy
    ]
})

results
```

# Hyperparameter

## Logistic Regression - Grid Search & Cross Validation

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid_lr = {
    "C": [0.01, 0.1, 1, 10, 100],
    "solver": ["liblinear", "lbfgs"]
}
```

```python
grid_lr = GridSearchCV(
    estimator=LogisticRegression(random_state=42),
    param_grid=param_grid_lr,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)
```

```python
grid_lr.fit(X_train, y_train)
```

```python
grid_lr.best_params_
```

```python
grid_lr.best_score_
```

```python
best_lr_model = grid_lr.best_estimator_
```

```python
y_val_pred_lr_tuned = best_lr_model.predict(X_val)
```

```python
tuned_lr_accuracy = accuracy_score(y_val, y_val_pred_lr_tuned)
print("Tuned Logistic Regression Validation Accuracy:", tuned_lr_accuracy)
```

```python
print(classification_report(y_val, y_val_pred_lr_tuned))
```

```python
grid_knn.best_params_
```

## Support Vector Machine (SVM) - Grid Search & Cross Validation

```python
param_grid_svm = {
    "C": [0.1, 1, 10],
    "gamma": ["scale"],
    "kernel": ["rbf"]
}
```

```python
grid_svm = GridSearchCV(
    estimator=SVC(random_state=42),
    param_grid=param_grid_svm,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)
```

```python
grid_svm.fit(X_train, y_train)
```

```python
grid_svm.best_params_
```

```python
best_svm_model = grid_svm.best_estimator_
```

```python
y_val_pred_svm_tuned = grid_svm.best_estimator_.predict(X_val)
```

```python
tuned_svm_accuracy = accuracy_score(y_val, y_val_pred_svm_tuned)
print("SVM Validation Accuracy:", tuned_svm_accuracy)
```

```python
tuned_results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "KNN",
        "SVM"
    ],
    "Tuned Validation Accuracy": [
        tuned_lr_accuracy,
        tuned_knn_accuracy,
        tuned_svm_accuracy
    ]
})

tuned_results
```

```python
y_test_pred_svm = grid_svm.best_estimator_.predict(X_test)
```

```python
test_svm_accuracy = accuracy_score(y_test, y_test_pred_svm)
print("Final SVM Test Accuracy:", test_svm_accuracy)
```
